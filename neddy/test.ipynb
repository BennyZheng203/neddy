```python
import requests
from astropy.io.votable import parse_single_table
from io import BytesIO
import pandas as pd
import matplotlib.pyplot as plt
# Define the target object
target_name = '3C 279'

# Construct the URL
url = f'https://vo.ned.ipac.caltech.edu/services/accessSED?REQUEST=getData&TARGETNAME={target_name}'

# Send the GET request
response = requests.get(url)

# Check if the request was successful
if response.status_code == 200:
    # Parse the VOTable data
    try:
        votable = parse_single_table(BytesIO(response.content))
    except IndexError:
        print(f'No Photometry/SED Data')
    data = votable.to_table()
else:
    print(f'Error: {response.status_code}')
```

```python
df = data.to_pandas()
```

```python
import requests
import xml.etree.ElementTree as ET
import pandas as pd

# Step 1: Fetch the XML data from the URL
url = "http://nedwww.ipac.caltech.edu/cgi-bin/nph-datasearch?search_type=Photometry&objname=3C+279&of=xml_main"
response = requests.get(url)

# Step 2: Parse the XML content
root = ET.fromstring(response.content)

# Step 3: Extract data from the TABLEDATA section
data_rows = []
for tabledata in root.findall(".//TABLEDATA"):
    for tr in tabledata.findall("TR"):
        row = [td.text.strip() if td.text else None for td in tr.findall("TD")]
        data_rows.append(row)

# Step 4: Extract column headers from FIELD elements
columns = [field.get("name") for field in root.findall(".//FIELD")]

# Step 5: Create a DataFrame
df = pd.DataFrame(data_rows, columns=columns)

# Display the DataFrame
df.head()
```

```python
import urllib
import requests
from fundamentals.download import multiobject_download
from astropy.io.votable import parse_single_table
from io import BytesIO
import logging as log
import pandas as pd

log.basicConfig(level=log.INFO)

def build_sed_query_urls(target_names):
    base_url = "https://vo.ned.ipac.caltech.edu/services/accessSED?REQUEST=getData&TARGETNAME="
    return [base_url + urllib.parse.quote(name) for name in target_names]

# List of target objects
target_names = ['3C 279', 'NGC 1275', 'M31']

# Build URLs for SED search
url_list = build_sed_query_urls(target_names)

# Download directory for SED data
download_directory = "/tmp"

# Download SED data using multiobject_download
sed_files = multiobject_download(
    urlList=url_list,
    downloadDirectory=download_directory,
    log=log,
    timeStamp=True,
    timeout=180,
    concurrentDownloads=5,
    resetFilename=False,
    credentials=False,
    longTime=True
)
```

```python
# Parse and display SED data
for file in sed_files:
    with open(file, 'rb') as f:
        try:
            votable = parse_single_table(BytesIO(f.read()))
            data = votable.to_table()
            df = data.to_pandas()
            df = df[(df['DataSpectralValue'] > 430 * 10**12) & (df['DataSpectralValue'] < 800 * 10**12)]
            print(df[['DataSpectralValue', 'DataFluxValue']].nlargest(3, 'DataFluxValue'))
            
        except Exception as e:
            print(f"Failed to parse SED data from {file}: {e}")
        
```
